```python
import nltk
from nltk.tokenize import word_tokenize
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
# Word Tokenization

text="Haris loves AI, Machine Learning, and Cricket ."
tokens=word_tokenize(text)
print(tokens)
```

```python
# Sentence Tokenization

text="""
My name is Haris Azeem.
I love coding In the Field Of Artifical intelligence.
AI Is changing the world.
With the help of NLP computer understand the language.
"""

from nltk.tokenize import sent_tokenize

text=text.strip()
text=text.lower()
token=sent_tokenize(text)
print(token)
```

```python
# Word Split and Sentence
text="""
My name is Haris Azeem.
I love coding In the Field Of Artifical intelligence.
AI Is changing the world.
With the help of NLP computer understand the language.
"""
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize

text1=text.lower()
text1=text.strip()

text=text.split()
token=word_tokenize(text1)
token=sent_tokenize(text1)

print(token)
```

```python
# Character Tokenization

text="Artifical"
char=list(text)
print("character",(char))
print("Total",len(char))
```
