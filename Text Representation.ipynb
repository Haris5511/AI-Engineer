```python
from sklearn.feature_extraction.text import CountVectorizer

docuements=[
    "I LOVE AI",
    "I LOVE PYTHON",
    "PYTHON IS AMAZING"
]

# Object Bannana

vector=CountVectorizer()
X=vector.fit_transform(docuements)

print(vector.get_feature_names_out())
print(X.toarray())
```

```python
text="I absolutely love this smartphone the camera is amazing the battery lasts all day and the performance is very fast however the price is a little high"

import re
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer

text=text.lower()
print(text)
token=word_tokenize(text)
print(token)

stopword=stopwords.words("english")
print(stopword)
filtered=[]
for words in token:
    if words not in stopword:
        filtered.append(words)
print(filtered)

docs=[
    "my name is Haris",
    "i am 23 years old"
]
count=CountVectorizer()
x=count.fit_transform(docs)
print(x.toarray())
print(count.get_feature_names_out())
```

```python
sent=[
    "i love coding",
    "its my passion",
    "Future is AI"
]
set=CountVectorizer()
x=set.fit_transform(sent)
print(x.toarray())
print(set.get_feature_names_out())
```

```python
# TF IDF 

from sklearn.feature_extraction.text import TfidfVectorizer
docs=[
    "I love machine learning and Python",
    "Python is used for machine learning projects", 
    "I love deep learning with neural networks",
    "Machine learning is very powerful"
]
vector=TfidfVectorizer()
x=vector.fit_transform(docs)
print(vector.get_feature_names_out())
print(x.toarray())

```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
test=[
    "deep learning models improve computer vision tasks",
    "computer vision uses deep neural networks",
    "natural language processing uses neural networks",
    "deep learning and neural networks are powerful",
    "computer vision and machine learning"
]
tf=TfidfVectorizer()
x=tf.fit_transform(test)
print(tf.get_feature_names_out())
print(x.toarray())
```
